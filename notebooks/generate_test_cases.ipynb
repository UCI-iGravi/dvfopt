# Generate Test Cases
Generate and save deformation fields for all experiment test cases.
Uses `testcases` for case definitions and `dvfopt.viz` for plotting.

## Imports

```python
import os
import numpy as np

from dvfopt.testdata import (
    SYNTHETIC_CASES, RANDOM_DVF_CASES, REAL_DATA_SLICES,
    make_deformation, make_random_dvf, load_slice, save_and_summarize,
)
from dvfopt.viz import plot_deformation_field
```

```python
from pathlib import Path

# Anchor OUTPUT_DIR to the project root so the notebook writes to
# <repo>/data/test_cases regardless of the kernel's current working
# directory (running from ./notebooks would otherwise create
# ./notebooks/data/test_cases).
_here = Path.cwd().resolve()
for _p in [_here, *_here.parents]:
    if (_p / "pyproject.toml").exists():
        PROJECT_ROOT = _p
        break
else:
    PROJECT_ROOT = _here

OUTPUT_DIR = str(PROJECT_ROOT / "data" / "test_cases")
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"Writing to: {OUTPUT_DIR}")
```

## Case 1 — Small synthetic grids (correspondences)

```python
for key in ["01a_10x10_crossing", "01b_10x10_opposite", "01c_20x40_edges", "01d_20x40_crossing"]:
    case = SYNTHETIC_CASES[key]
    deformation, ms, fs = make_deformation(key)
    save_path = f"{OUTPUT_DIR}/{key}.npy"
    save_and_summarize(deformation, save_path)
    plot_deformation_field(deformation, ms, fs, title=case["title"],
                           show_values=False, show_points=False, save_path=save_path)
```

## Case 1e–f — Random DVFs

```python
for key in ["01e_20x20_random_spirals", "01f_20x20_random_seed_42"]:
    case = RANDOM_DVF_CASES[key]
    dvf = make_random_dvf(key)
    save_path = f"{OUTPUT_DIR}/{key}.npy"
    save_and_summarize(dvf, save_path)
    plot_deformation_field(dvf, title=case["title"],
                           show_values=False, show_points=False, save_path=save_path)
```

## Case 2 — Real data slices

```python
mpoints_path = str(PROJECT_ROOT / "data" / "corrected_correspondences_count_touching" / "mpoints.npy")
fpoints_path = str(PROJECT_ROOT / "data" / "corrected_correspondences_count_touching" / "fpoints.npy")

for key, cfg in REAL_DATA_SLICES.items():
    deformation, ms, fs = load_slice(
        cfg["slice_idx"], scale_factor=cfg["scale_factor"],
        mpoints_path=mpoints_path, fpoints_path=fpoints_path,
    )
    save_path = f"{OUTPUT_DIR}/{key}_slice{cfg['slice_idx']}.npy"
    save_and_summarize(deformation, save_path)
    plot_deformation_field(deformation, ms, fs, title=cfg["title"],
                           show_values=False, show_points=False,
                           save_path=save_path, quiver_scale=1)
```

## Case 3 — Additional synthetic grids

```python
for key in ["03a_10x10_opposite", "03b_10x10_crossing", "03c_20x20_opposite", "03d_20x20_crossing"]:
    case = SYNTHETIC_CASES[key]
    deformation, ms, fs = make_deformation(key)
    save_path = f"{OUTPUT_DIR}/{key}.npy"
    save_and_summarize(deformation, save_path)
    plot_deformation_field(deformation, ms, fs, title=case["title"],
                           show_values=False, show_points=False, save_path=save_path)
```

## Checkerboard test case

```python
# Create and save a checkerboard deformation field
save_path = f"{OUTPUT_DIR}/checkerboard/phi.npy"

deformation_cb = np.zeros((3, 1, 10, 10), dtype=np.float32)
deformation_cb[2, 0, :, :-1] = 1
for row in [1, 3, 5, 7]:
    deformation_cb[2, 0, row, 3] = -2
    deformation_cb[2, 0, row, 7] = -2

os.makedirs(os.path.dirname(save_path), exist_ok=True)
save_and_summarize(deformation_cb, save_path)
plot_deformation_field(deformation_cb, title="Checkerboard deformation (10×10)",
                       show_values=False, show_points=False, save_path=save_path)
```
